# Email Column

```python
# Let's write a code to handle all the null values in the email column using the pattern
def handle_null_email(name):
    first, last = name.split()
    return f"{first.lower()}.{last.lower()}@gmail.com"
```

```python
df['Email'] = df.apply(
    lambda x: handle_null_email(x['Name']) if pd.isna(x['Email']) else x['Email'],
    axis=1
)
```
